> ### Status Codes

> 🟢 **2xx Success** — Request processed as expected  
> 🟡 **4xx Client Error** — Issue with the incoming request  
> 🔴 **5xx Server Error** — Issue on the server side 


##### 🟢 Success (2xx)

| Status | Code | Description
| :--- | :--- | :--- |
| **OK** | `200` | The request succeeded.
| **Created** | `201` | A new resource was successfully created.
| **Accepted** | `202` | Request accepted for asynchronous processing. 
| **No Content** | `204` | Request succeeded, no body returned *(e.g. `DELETE`)*.


##### 🟡 Client Errors (4xx)

| Status | Code | Cause & Remediation 
| :--- | :--- | :--- |
| **Bad Request** | `400` | Invalid request syntax or missing required fields.
| **Unauthorized** | `401` | Authentication token is missing or invalid.
| **Forbidden** | `403` | Valid credentials, but lacks access to resource.
| **Not Found** | `404` | Requested endpoint or resource does not exist.
| **Unprocessable** | `422` | Request structure valid, but failed domain validation.
| **Too Many Requests** | `429` | Rate limit exceeded. Retry after designated window.


##### 🔴 Server Errors (5xx)

| Status | Code | Description 
| :--- | :--- | :--- |
| **Internal Error** | `500` | Unexpected failure on the server side.
| **Bad Gateway** | `502` | Upstream service or reverse proxy failed. logs.
| **Service Unavailable**| `503` | Server temporarily overloaded or down for maintenance.
| **Gateway Timeout** | `504` | Upstream service failed to respond in time.
