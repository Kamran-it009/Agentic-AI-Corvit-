```python
import json
```

```python
python_dict = {10: 'Kamran', 20: 'Ali', 30: 'Kashif'}
print(python_dict)
```

### Dictionary to JSON Conversion

```python
json_obj = json.dumps(python_dict)
print(json_obj)
print(type(json_obj))
```

### JSON to Dictionary Conversion

```python
dict_2 = json.loads(json_obj)
print(dict_2)
print(type(dict_2))
```
