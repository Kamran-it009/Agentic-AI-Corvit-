### The Power of Pydantic

```python
from pydantic import BaseModel, Field


# 1. Define your data schema and validation rules
class User(BaseModel):
    name: str
    age: int
```

```python
u1 = User(name='Kamran', age=30)  # Valid
print(u1)
```

```python
u2 = User(name="Kamran", age='thirty')  # Invalid, raises ValidationError due to min_length constraint
print(u2)
```

#### Data validation for Sequence Types

```python
class Person(BaseModel):
    name: str
    age: int
    hobbies: list[str]

user = Person(name="Kamran", age=30, hobbies=["reading", "coding"])
print(user)
```

#### Typing module

```python
from typing import List

class Person(BaseModel):
    name: str
    age: int
    hobbies: List[str]

user = Person(name="Kamran", age=30, hobbies=["reading", "coding"])
print(user)
```

```python
class Person(BaseModel):
    name: str
    age: int
    hobbies: list[str]

user = Person(name="Kamran", age=30, hobbies=["reading", 20])
print(user)
```

### Optional Data

```python
from typing import Optional

class Person(BaseModel):
    name: str
    age: int
    hobbies: List[str]
    # books: Optional[List[str]] = None
    books: list[str] | None = None

user = Person(name="Kamran", age=30, hobbies=["reading", "coding"])
print(user)
```

### Built-in validators in Pydantic

```python
class Person(BaseModel):
    name: str
    age: int
    email: str

user = Person(name="Kamran", age=30, email="kamran_example.com")
print(user)
```

```python
from pydantic import BaseModel, EmailStr

class Person(BaseModel):
    name: str
    age: int
    email: EmailStr

user = Person(name="Kamran", age=30, email="kamran_example.com")
print(user)
```

#### Use of `Field` for additional validation

```python
from pydantic import BaseModel, Field
class Person(BaseModel):
    name: str
    age: int
    weight: float = Field(gt=0, lt = 50)

user = Person(name="Kamran", age=30, weight=70.5)
print(user)
```

```python
from pydantic import BaseModel, Field

class Person(BaseModel):
    name: str = Field(min_length=3, max_length=20)
    age: int
    weight: float = Field(gt=0, lt = 50)

user = Person(name="Kamran", age=30, weight=40)
print(user)
```

#### Model Methods and Properties

```python
class User(BaseModel):
    name: str
    age: int = Field(strict=True)
```

```python
# Validates a dictionary/object
data_dict = {"name": "Alice", "age": 30}
user1 = User.model_validate(data_dict)
print(user1)
```

```python
#  Converts model into a Python dictionary
data = user1.model_dump()
print(data)
print(type(data))
```

```python
# Converts model into a JSON string
data = user1.model_dump_json()
print(data)
```
