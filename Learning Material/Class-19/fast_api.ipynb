# Fast API
FastAPI is a modern, fast (high-performance), web framework for building APIs with Python based on standard Python type hints. [See documentation](https://fastapi.tiangolo.com/)

### Step-1: Installation
```uv add "fastapi[standard]"```

### Step-2: 

Create a file with name ```main.py``` and Write the following code

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
async def root():
    return {"message": "Hello World"}
````

### Step-3: Run the development server
```uv run fastapi dev```

**INFO:     Uvicorn running on http://127.0.0.1:8000 (Press CTRL+C to quit)**

*You may also save the file with different name, In this case, you will run the following command to start the server:*

```uv run uvicorn file_name:app```

## 1. Path Parameters 

````python
from fastapi import FastAPI

app = FastAPI()


@app.get("/products")
async def read_item():
    return {"List of products": ['Laptop', 'Mobile', 'PC', 'Head-phones']}
````

````python
from fastapi import FastAPI

app = FastAPI()

products = {1: "Laptop", 2: "Mobile", 3: "PC", 4: "Head-phones"}

@app.get("/products/{product_id}")
async def read_item(product_id: int):
    p = products.get(product_id)
    return {"product_id": product_id, "product_name": p}
````
